```python
INPUT_DIRS = [
    "/work/scratch-nopw2/mendrika/pancast/raw/inputs_t0",
    "/gws/nopw/j04/wiser_ewsa/mrakotomanga/pancast/raw/inputs_t0",
]

all_inputs = []

for d in INPUT_DIRS:
    if not os.path.exists(d):
        continue
    all_inputs.extend(glob(os.path.join(d, "input-*.pt")))

all_inputs = sorted(set(all_inputs))

print("Total inputs found:", len(all_inputs))
```

```python
def extract_year(filename):
    """
    Extracts 4-digit year from filenames like:
    input-20181230_1200.pt → 2018
    """
    match = re.search(r"(\d{4})\d{2}\d{2}_\d{4}", filename)
    if match:
        return int(match.group(1))
    else:
        return None
```

```python
train_inputs = []
val_inputs = []
test_inputs = []

for f in all_inputs:
    year = extract_year(f)

    if 2004 <= year <= 2022:
        train_inputs.append(f)
    elif year == 2023:
        val_inputs.append(f)
    elif year == 2024:
        test_inputs.append(f)
```

```python
print("Train:", len(train_inputs))
print("Val:  ", len(val_inputs))
print("Test: ", len(test_inputs))

assert not set(train_inputs) & set(val_inputs)
assert not set(train_inputs) & set(test_inputs)
assert not set(val_inputs) & set(test_inputs)
```

```python
import json

splits = {
    "train": train_inputs,
    "val": val_inputs,
    "test": test_inputs,
}

with open("/home/users/mendrika/Object-Based-LSTMConv/outputs/data-split/input_splits.json", "w") as f:
    json.dump(splits, f, indent=2)
```
